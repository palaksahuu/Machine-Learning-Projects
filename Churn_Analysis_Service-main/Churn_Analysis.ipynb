```python
# convet to or 1 to yes or no for senoircitizen column
def conv(value):
    if value ==1 :
        return "yes"
    else:
        return "no"

df['SeniorCitizen'] = df["SeniorCitizen"].apply(conv)
```

```python

# Define categorical columns
categorical_columns = [
    'gender', 'SeniorCitizen', 'Partner', 'Dependents', 'PhoneService', 
    'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup', 
    'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 
    'Contract', 'PaperlessBilling', 'PaymentMethod', 'Churn'
]

# Set up subplot grid size
num_cols = len(categorical_columns)
rows = (num_cols // 3) + (num_cols % 3 > 0)  # Arrange in a 3-column layout

fig, axes = plt.subplots(nrows=rows, ncols=3, figsize=(18, rows * 5))
axes = axes.flatten()  # Flatten axes array for easy indexing

# Loop through categorical columns and plot countplots
for i, col in enumerate(categorical_columns):
    ax = axes[i]
    sns.countplot(x=df[col], data=df, ax=ax, palette="coolwarm" , hue = "Churn")
    ax.set_title(f"Count of {col}", fontsize=12)
    ax.set_xlabel(col, fontsize=10)
    ax.set_ylabel("Count", fontsize=10)
    ax.tick_params(axis='x', rotation=30)  # Rotate x-axis labels for readability

# Remove empty subplots if any
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Adjust layout
plt.tight_layout()
plt.show()
```
